## Ejercicio 1: Lista de Tareas con Lista Enlazada

**Título:** Gestión de una lista de tareas pendientes mediante una lista
enlazada

```python
# El nodo para la lista enlazada
class nodo_nuevaTarea:
    def __init__(self, nombre_tarea, estado_actual=True):
        self.nombre_tarea = nombre_tarea
        self.estado_actual = estado_actual # Donde True = pendiente, False = completada
        self.siguiente = None         # Puntero para el siguiente nodo


class lista_tareas:

  # Inicio de la lista
    def __init__(self):
        self.cabeza = None

  # verifica si esta vacia
    def esta_vacia(self):
        return self.cabeza is None

  # para insertar una tarea al inicio de la lista
    def insertar_inicio(self, nombre_tarea, estado_actual=True):
        nuevo = nodo_nuevaTarea(nombre_tarea, estado_actual)
        nuevo.siguiente = self.cabeza
        self.cabeza = nuevo

  # Elimina solo la primera tarea con ese nombre
    def eliminar_tarea(self, nombre_tarea):
        actual = self.cabeza
        anterior = None

        # Buscar la tarea
        while actual is not None and actual.nombre_tarea != nombre_tarea:
            anterior = actual
            actual = actual.siguiente

        # Devuelve False si la tarea no se encontró
        if actual is None:
            return False

        # Devuelve True si la tarea se elimino
        if anterior is None:
            self.cabeza = actual.siguiente
        else:
            anterior.siguiente = actual.siguiente

        return True

  # Solo busca una tarea por nombre y retorna su nodo o "None si la tarea no existe"
    def buscar_tarea(self, nombre_tarea):
        actual = self.cabeza
        while actual is not None:
            if actual.nombre_tarea == nombre_tarea:
                return actual
            actual = actual.siguiente
        return None

  # Imprime todas las tareas registradas
    def mostrar_tarea(self):
        if self.esta_vacia():
            print("La lista de tareas está vacía.")
            return

        actual = self.cabeza
        print("Tareas registradas:")
        while actual is not None:
            estado_actual_texto = "Pendiente" if actual.estado_actual else "Completada"
            print(f"- {actual.nombre_tarea} ({estado_actual_texto})")
            actual = actual.siguiente

  # Marca como completada la tarea indicada
    def marcar_completada(self, nombre_tarea):
        tarea = self.buscar_tarea(nombre_tarea)
        if tarea is not None:
            tarea.estado_actual = False
            return True
        return False
```

```python
def menu_lista_tareas():
    lista = lista_tareas()
  # Se imprime el menu para que el usuario elija que quiere hacer
    while True:
        print("\n   MENÚ LISTA DE TAREAS")
        print("1. Insertar tarea al inicio")
        print("2. Eliminar tarea por nombre")
        print("3. Buscar tarea")
        print("4. Marcar tarea como completada")
        print("5. Mostrar todas las tareas")
        print("6. Salir")
        opcion = input("Elige una opción: ")

        if opcion == "1":
            nombre = input("Nombre de la tarea: ")
            lista.insertar_inicio(nombre, True)
        elif opcion == "2":
            nombre = input("Nombre de la tarea a eliminar: ")
            if lista.eliminar_tarea(nombre):
                print("Tarea eliminada.")
            else:
                print("Tarea no encontrada.")
        elif opcion == "3":
            nombre = input("Nombre de la tarea a buscar: ")
            tarea = lista.buscar_tarea(nombre)
            if tarea is not None:
                estado_actual = "Pendiente" if tarea.estado_actual else "Completada"
                print(f"Tarea encontrada: {tarea.nombre_tarea} ({estado_actual})")
            else:
                print("Tarea no encontrada.")
        elif opcion == "4":
            nombre = input("Nombre de la tarea a marcar como completada: ")
            if lista.marcar_completada(nombre):
                print("Tarea marcada como completada.")
            else:
                print("Tarea no encontrada.")
        elif opcion == "5":
            lista.mostrar_tarea()
        elif opcion == "6":
            print("Saliendo del sistema de tareas...")
            break #Break para salir del menu
        else:
            print("Opción inválida. Intenta de nuevo.") # Verifica que la opcion sea valida

# Para ejecutar el menu
menu_lista_tareas()
```

# Evaluación Parcial I - Estructura de Datos
## Ejercicio 2: Simulación de Farmacia con Pila y Cola (Listas Enlazadas)

Este notebook implementa una **Cola** para clientes y una **Pila** para pedidos (medicamentos) usando listas enlazadas en Python, con un menú interactivo para probar en Google Colab.

```python
# El nodo para la lista enlazada
class Nodo:
    def __init__(self, dato):
        self.dato = dato
        self.siguiente = None

#Clase cola para manejar los clientes
class Cola:
    def __init__(self):
        self.frente = None
        self.final = None

#Verificar si esta vacia
    def esta_vacia(self):
        return self.frente is None

#Agrega un elemento al final de la cola.
    def encolar_cliente(self, dato):
        nuevo = Nodo(dato)
        if self.esta_vacia():
            self.frente = self.final = nuevo
        else:
            self.final.siguiente = nuevo
            self.final = nuevo
  #Elimina y retorna el elemento del frente. Retorna None si la cola está vacía
    def desencolar_cliente(self):
        if self.esta_vacia():
            return None

        dato = self.frente.dato
        self.frente = self.frente.siguiente
        if self.frente is None:
            self.final = None
        return dato

  #Muestra el estado_actual actual de la cola
    def mostrar_cliente(self):
        if self.esta_vacia():
            print("Cola de clientes vacía.")
            return

        actual = self.frente
        print("Clientes en cola ( de frente al final):")
        while actual is not None:
            print(f"- {actual.dato}")
            actual = actual.siguiente
```

```python
class Pila:
    def __init__(self):
        self.tope = None    # Último elemento apilado

  #Verificar si esta vacia
    def esta_vacia(self):
        return self.tope is None

  # Ingresar nuevo dato al tope de la pila
    def apilar(self, dato):
        nuevo = Nodo(dato)
        nuevo.siguiente = self.tope
        self.tope = nuevo

  # Elimina y retorna el elemento del tope. Retorna None si la pila está vacía
    def desapilar(self):
        if self.esta_vacia():
            return None

        dato = self.tope.dato
        self.tope = self.tope.siguiente
        return dato
  # Muestra el estado actual de la pila
    def mostrar_pila(self):
        if self.esta_vacia():
            print("Pila de pedidos vacía.")
            return

        actual = self.tope
        print("Pedidos preparados en pila (del tope al fondo):")
        while actual is not None:
            print(f"- {actual.dato}")
            actual = actual.siguiente
```

```python
def menu_farmacia():
    cola_clientes = Cola()
    pila_pedidos = Pila()

  # Se imprime el menu para que el usuario elija que quiere hacer
    while True:
        print("\n  MENÚ DE FARMACIA ")
        print("1) Nuevo cliente")
        print("2) Atender cliente (desencolar)")
        print("3) Nuevo pedido de medicamento (apilar)")
        print("4) Entregar último medicamento preparado (desapilar)")
        print("5) Mostrar el estado de la cola y la pila")
        print("6) Salir")
        opcion = input("Elige una opción: ")

        if opcion == "1":
            nombre = input("Nombre del cliente: ")
            cola_clientes.encolar_cliente(nombre)
            print(f"Cliente {nombre} encolado.")
        elif opcion == "2":
            atendido = cola_clientes.desencolar_cliente()
            if atendido is None:
                print("No hay clientes en la cola.")
            else:
                print(f"Cliente atendido: {atendido}")
        elif opcion == "3":
            medicamento = input("Nombre del medicamento preparado: ")
            pila_pedidos.apilar(medicamento)
            print(f"Pedido '{medicamento}' apilado.")
        elif opcion == "4":
            entregado = pila_pedidos.desapilar()
            if entregado is None:
                print("No hay pedidos preparados en la pila.")
            else:
                print(f"Medicamento entregado: {entregado}")
        elif opcion == "5":
            print("\nEstado actual de la farmacia:")
            cola_clientes.mostrar_cliente()
            pila_pedidos.mostrar_pila()
        elif opcion == "6":
            print("Saliendo del sistema de farmacia...")
            break #Break para salir del menu
        else:
            print("Opción inválida. Intenta nuevamente.") # Verifica que la opcion sea valida

# Para ejecutar el menu
menu_farmacia()
```
